## 1. Cálculo Manual de Similaridade de Cosseno
$$\text{Cosine Similarity}(A, B) = \frac{A \cdot B}{\|A\| \|B\|}$$

```python
def cosine_similarity(v1, v2):
    dot_product = np.dot(v1, v2)
    norm_v1 = np.linalg.norm(v1)
    norm_v2 = np.linalg.norm(v2)
    return dot_product / (norm_v1 * norm_v2)

# Exemplo conceitual com vetores tridimensionais
v_python = np.array([0.9, 0.1, 0.8])
v_code   = np.array([0.85, 0.15, 0.75])
v_pizza  = np.array([0.05, 0.95, 0.10])

print(f"Similaridade ('Python', 'Código'): {cosine_similarity(v_python, v_code):.4f}")
print(f"Similaridade ('Python', 'Pizza'):  {cosine_similarity(v_python, v_pizza):.4f}")
```

## 2. In-Memory Vector Store e Busca Top-K

```python
np.random.seed(42)

# Simulando uma base de conhecimento com 5 documentos e vetores de dimensão 8
documentos = [
    "Configuração de rotas de rede e VLANs corporativas",
    "Desenvolvimento de APIs RESTful com Node.js e Express",
    "Monitoramento de tráfego de rede e análise de pacotes",
    "Modelagem de banco de dados relacional com PostgreSQL",
    "Instalação e gerenciamento de servidores Linux Ubuntu"
]

# Gerando embeddings simulados (normalizados)
def mock_embedding_generator(dim=8):
    v = np.random.randn(dim)
    return v / np.linalg.norm(v)

vector_db = {doc: mock_embedding_generator() for doc in documentos}

# Query de busca do usuário
query = "Como analisar o tráfego de redes e pacotes?"
query_vector = mock_embedding_generator() # Vetor da consulta

# Cálculo de similaridade com toda a base
resultados = []
for doc, doc_vector in vector_db.items():
    score = cosine_similarity(query_vector, doc_vector)
    resultados.append((doc, score))

# Ordenação Top-K (maior similaridade primeiro)
resultados.sort(key=lambda x: x[1], reverse=True)

print(f"🔍 Consulta: '{query}'\n")
print("Top Documentos Relevantes:")
for idx, (doc, score) in enumerate(resultados[:3], 1):
    print(f"{idx}. [{score:.4f}] {doc}")
```
