## Objetivos
1. Compreender a diferença entre **Zero-Shot** e **Few-Shot Prompting**.
2. Implementar a técnica **Chain-of-Thought (CoT)** para raciocínio passo a passo.
3. Garantir saídas estruturadas em formato **JSON** validadas programaticamente.

```python
import json
import re

def simular_resposta_llm(prompt_template):
    """
    Função simuladora para demonstrar a estrutura dos prompts e respostas esperadas.
    """
    print("=== PROMPT ENVIADO AO MODELO ===")
    print(prompt_template.strip())
    print("=" * 32)

# Exemplo de System Prompt + User Prompt
system_prompt = "Você é um especialista em cibersegurança e análise de logs de rede."
user_prompt = "Analise o seguinte IP suspeito: 192.168.1.105"

prompt_completo = f"SYSTEM: {system_prompt}\nUSER: {user_prompt}"
simular_resposta_llm(prompt_completo)
```

## 1. Zero-Shot vs. Few-Shot Prompting

```python
# Padrão Few-Shot (Exemplos no próprio contexto)
few_shot_prompt = """
Classifique a intenção do usuário em: [Suporte, Vendas, Cancelamento]

Exemplo 1:
Texto: "Gostaria de saber os preços do plano corporativo."
Intenção: Vendas

Exemplo 2:
Texto: "Meu sistema está apresentando erro 500 ao logar."
Intenção: Suporte

Texto: "Quero encerrar minha assinatura imediatamente."
Intenção:
"""

simular_resposta_llm(few_shot_prompt)
```

## 2. Chain-of-Thought (CoT) Prompting

```python
cot_prompt = """
Resolva o problema raciocinando passo a passo antes de dar a resposta final.

Pergunta:
Em uma rede local, temos 3 sub-redes com 20, 45 e 10 hosts. Quantos IPs úteis são necessários no mínimo, considerando o gateway de cada sub-rede?

Pense passo a passo:
1. Calcule os IPs de host para a Sub-rede 1 (20 hosts + 1 gateway).
2. Calcule os IPs de host para a Sub-rede 2 (45 hosts + 1 gateway).
3. Calcule os IPs de host para a Sub-rede 3 (10 hosts + 1 gateway).
4. Some os totais e determine a resposta final.
"""

simular_resposta_llm(cot_prompt)
```

## 3. Saída Estruturada em JSON com Validação

```python
json_prompt = """
Extraia as entidades do texto abaixo e retorne ESTRITAMENTE um objeto JSON válido.
Não inclua nenhum texto explicativo antes ou depois do JSON.

Formato esperado:
{
  "usuario": "nome",
  "servico": "nome do serviço",
  "severidade": "BAIXA | MEDIA | ALTA"
}

Texto:
O usuário Carlos Relato informou uma falha de instabilidade crítica no serviço de Banco de Dados PostgreSQL.
"""

# Resposta simulada em formato string JSON
resposta_modelo_raw = '{"usuario": "Carlos Relato", "servico": "Banco de Dados PostgreSQL", "severidade": "ALTA"}'

# Validação do JSON
try:
    dados_extraidos = json.loads(resposta_modelo_raw)
    print("✅ JSON Parseado com sucesso:")
    print(json.dumps(dados_extraidos, indent=2, ensure_ascii=False))
except json.JSONDecodeError as e:
    print("❌ Falha na validação do JSON:", e)
```
