## Objetivos
1. Modelar agentes orientados a objetos com System Prompts e funções especializadas.
2. Construir um pipeline sequencial onde a saída de um agente serve como entrada para o próximo.
3. Gerenciar o estado global da conversa e o histórico de mensagens entre agentes.

```python
from dataclasses import dataclass, field
from typing import List, Dict

@dataclass
class Message:
    sender: str
    content: str

@dataclass
class Agent:
    name: str
    role: str
    system_prompt: str
    
    def process(self, input_text: str, context: List[Message]) -> str:
        """
        Simula o processamento do LLM para este papel específico, 
        incorporando a mensagem de entrada e o contexto.
        """
        prompt = f"[{self.role.upper()} - {self.name}]\nSYSTEM: {self.system_prompt}\nINPUT: {input_text}"
        
        # Respostas simuladas com base no papel
        if self.role == "researcher":
            return f"Fatos encontrados: 1. O protocolo P4 permite programar o plano de dados. 2. BMv2 é a arquitetura de software de referência para P4."
        elif self.role == "analyst":
            return f"Análise técnica: O P4 reduz o processamento no plano de controle delegando classificação de pacotes ao hardware/ASIC."
        elif self.role == "writer":
            return f"### Resumo Executivo\nSua tecnologia utiliza P4_16 para classificação eficiente no data plane, resultando em altíssimo throughput."
        
        return "Processado sem regras específicas."
```

## 1. Instanciação do Time de Agentes Especializados

```python
agente_pesquisador = Agent(
    name="Agente-Busca",
    role="researcher",
    system_prompt="Você é um especialista em busca técnica. Sua função é extrair fatos objetivos sobre a demanda."
)

agente_analista = Agent(
    name="Agente-Arquitetura",
    role="analyst",
    system_prompt="Você é um arquiteto de sistemas. Sua função é analisar criticamente os fatos fornecidos e identificar implicações técnicas."
)

agente_redator = Agent(
    name="Agente-Documentador",
    role="writer",
    system_prompt="Você é um redator técnico. Sua função é sintetizar análises técnicas em um formato claro e pronto para publicação."
)

time_de_agentes = [agente_pesquisador, agente_analista, agente_redator]
print(f"👥 Equipe montada com {len(time_de_agentes)} agentes especializados.")
```

## 2. Pipeline de Execução Sequencial (Sequential Multi-Agent Workflow)

```python
def execute_multi_agent_pipeline(objective: str, agents: List[Agent]):
    shared_memory: List[Message] = []
    current_input = objective
    
    print(f"🚀 Iniciando Workflow Multi-Agente")
    print(f"🎯 Objetivo Inicial: '{objective}'\n")
    print("=" * 60)
    
    for agent in agents:
        output = agent.process(current_input, shared_memory)
        
        # Registra a mensagem na memória compartilhada
        msg = Message(sender=f"{agent.name} ({agent.role})", content=output)
        shared_memory.append(msg)
        
        print(f"📌 [{msg.sender}]")
        print(f"{msg.content}\n")
        print("-" * 60)
        
        # A saída do agente atual torna-se a entrada do próximo
        current_input = output
        
    return shared_memory

# Executando o pipeline com um objetivo de rede
resultado_final = execute_multi_agent_pipeline(
    objective="Explicar os benefícios do processamento de tráfego de rede com P4_16",
    agents=time_de_agentes
)
```
