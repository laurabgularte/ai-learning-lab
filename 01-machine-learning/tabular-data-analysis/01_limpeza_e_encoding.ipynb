## Objetivos
1. Identificar e tratar **valores ausentes** (NaN) usando estratégias de imputação.
2. Codificar variáveis categóricas com **One-Hot Encoding** e **Ordinal Encoding**.
3. Detectar e tratar **outliers** utilizando a técnica do Intervalo Interquartil (IQR).
4. Construir um **Pipeline de Pré-processamento** reutilizável com `ColumnTransformer`.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# 1. Criação de um dataset sintético com dados ausentes e categóricos
data = {
    'idade': [25, 30, np.nan, 45, 22, 50, 35, np.nan],
    'salario': [50000, 60000, 52000, 110000, 48000, 125000, np.nan, 62000],
    'escolaridade': ['Médio', 'Superior', 'Superior', 'Pós', 'Médio', 'Pós', 'Superior', 'Médio'],
    'departamento': ['TI', 'HR', 'TI', 'Financeiro', 'HR', 'Financeiro', 'TI', 'HR'],
    'churn': [0, 0, 1, 0, 1, 0, 0, 1]
}

df = pd.DataFrame(data)
print("--- Dataset Bruto ---")
display(df)
```

## Pipeline de Pré-processamento Automatizado

```python
# Separação dos atributos por tipo
num_cols = ['idade', 'salario']
cat_nominal_cols = ['departamento']
cat_ordinal_cols = ['escolaridade']

# 1. Pipeline para variáveis numéricas (Imputação pela Mediana + Padronização)
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# 2. Pipeline para variáveis nominais (Imputação Moda + OneHotEncoder)
cat_nominal_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(drop='first', sparse_output=False))
])

# 3. Pipeline para variáveis ordinais
ordem_escolaridade = [['Médio', 'Superior', 'Pós']]
cat_ordinal_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('ordinal', OrdinalEncoder(categories=ordem_escolaridade))
])

# Combinação de todos os pipelines
preprocessor = ColumnTransformer(transformers=[
    ('num', num_pipeline, num_cols),
    ('cat_nom', cat_nominal_pipeline, cat_nominal_cols),
    ('cat_ord', cat_ordinal_pipeline, cat_ordinal_cols)
])

# Aplicação da transformação nos dados
X = df.drop(columns=['churn'])
y = df['churn']

X_processed = preprocessor.fit_transform(X)
print("✅ Dados pré-processados e prontos para modelo!")
print("Formato da matriz resultante:", X_processed.shape)
```
